Being that we are interested in determining the habitability of exoplanets, it is important to consider factors
that mainly concern the planets themselves. These factors include the planet's distance from its star, its size
and mass, its atmospheric composition, and the presence of liquid water. 
We do not need to focus on many other features that are less directly related to the planet's habitability in our
dataset. If we want to efficiently clean the data, we should focus on cleaning null values from planet related
features specifically.

```python
# Keep unique planets and rows with a known target.
planetData = planetData.drop_duplicates()
target_column = "habitability_category"
feature_columns = [
    "pl_rade", "pl_masse", "pl_eqt", "pl_insol",
    "pl_orbsmax", "pl_orbper", "pl_orbeccen", "st_teff",
]
modelData = planetData.dropna(subset=[target_column]).copy()
X = modelData[feature_columns]
y = modelData[target_column]

print(f"Rows available: {len(X)}")
```

Now we use stratified k-fold cross validation and test with logistic regression. SMOTE generates synthetic samples of minority classes. It is placed inside an imblearn pipeline so it is only applied to each training fold (never the validation fold), avoiding data leakage.

```python
from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score, balanced_accuracy_score, f1_score
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler

print(y.value_counts())

# SMOTE needs k_neighbors < smallest class size within a training fold
n_splits = 5
min_class = y.value_counts().min()
k_neighbors = max(1, min(5, int(min_class * (n_splits - 1) / n_splits) - 1))

# The same folds are reused for every algorithm so the results are comparable
cv = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)
scoring = ["accuracy", "balanced_accuracy", "f1_macro"]

def run_kfold(model_pipeline, name):
    """Train on k-1 folds and test on the held-out fold, for every fold."""
    results = {f"test_{m}": [] for m in scoring}
    for fold, (train_idx, test_idx) in enumerate(cv.split(X, y), start=1):
        X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
        y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

        model_pipeline.fit(X_train, y_train)  # training (SMOTE only touches the training fold)
        y_pred = model_pipeline.predict(X_test)  # testing on unseen data

        results["test_accuracy"].append(accuracy_score(y_test, y_pred))
        results["test_balanced_accuracy"].append(balanced_accuracy_score(y_test, y_pred))
        results["test_f1_macro"].append(f1_score(y_test, y_pred, average="macro"))
        print(f"{name} fold {fold}: train={len(train_idx)} test={len(test_idx)} "
              f"accuracy={results['test_accuracy'][-1]:.3f}")

    results = {k: np.array(v) for k, v in results.items()}
    for metric in scoring:
        s = results[f"test_{metric}"]
        print(f"{metric}: {s.mean():.3f} +/- {s.std():.3f}")
    return results
```

Training 1: logistic regression. Each fold trains a fresh model on the training folds and tests it on the held-out fold.

```python
lr_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("smote", SMOTE(k_neighbors=k_neighbors, random_state=42)),
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(max_iter=2000, random_state=42)),
])

lr_scores = run_kfold(lr_pipeline, "LogisticRegression")
```

Training 2: decision tree. Same data, same stratified k-folds and same SMOTE step as logistic regression; each fold trains a fresh tree and tests it on the held-out fold, so the two results are directly comparable.

```python
from sklearn.tree import DecisionTreeClassifier

tree_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("smote", SMOTE(k_neighbors=k_neighbors, random_state=42)),
    ("model", DecisionTreeClassifier(max_depth=6, min_samples_leaf=3, random_state=42)),
])

tree_scores = run_kfold(tree_pipeline, "DecisionTree")
```

Random Forests:

scikit RandomForestClassifier documentation: https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.RandomForestClassifier.html

Random forest is a commonly-used machine learning algorithm, trademarked by Leo Breiman and Adele Cutler, that combines the output of multiple decision trees to reach a single result. Its ease of use and flexibility have fueled its adoption, as it handles both classification and regression problems.-
https://www.ibm.com/think/topics/random-forest#1989360318




```python
from sklearn.ensemble import RandomForestClassifier

rf_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("smote", SMOTE(k_neighbors=k_neighbors, random_state=42)),
    ("model", RandomForestClassifier(
        n_estimators=200,
        max_depth=None,
        min_samples_split=2,
        random_state=42,
        n_jobs=-1,
    )),
])

rf_scores = run_kfold(rf_pipeline, "RandomForest")
```

Support Vector Machines

Documentation: https://scikit-learn.org/stable/modules/svm.html

Paper on SVMs: https://homepages.math.uic.edu/%7Elreyzin/papers/cortes95.pdf

A support vector machine (SVM) is a supervised machine learning algorithm that classifies data by finding an optimal line or hyperplane that maximizes the distance between each class in an N-dimensional space. - https://www.ibm.com/think/topics/support-vector-machine#1840295164

```python
from sklearn.svm import SVC
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline

svm_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("smote", SMOTE(k_neighbors=k_neighbors, random_state=42)),
    ("model", SVC(
        kernel="rbf",
        C=1.0,
        gamma="scale",
    )),
])

svm_scores = run_kfold(svm_pipeline, "SVM")
```

Side-by-side comparison of all four algorithms.

```python
comparison = pd.DataFrame({
    "Logistic Regression": {m: lr_scores[f"test_{m}"].mean() for m in scoring},
    "Decision Tree": {m: tree_scores[f"test_{m}"].mean() for m in scoring},
    "Random Forest": {m: rf_scores[f"test_{m}"].mean() for m in scoring},
    "SVM": {m: svm_scores[f"test_{m}"].mean() for m in scoring},
}).round(3)
comparison
```
